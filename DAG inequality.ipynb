```python
import graphviz

COLORS = [
    '#00B0F0',   # blue
    '#FF0000',   # red
    '#B0F000'    # green
]
```

```python
graph = graphviz.Digraph(format='png', engine='neato')

nodes = ['D', 'X', 'Y']
positions = ['1,1!', '3,0!', '1, 0!']

edges = ['XY', 'DY']

[graph.node(n, pos=pos) for n, pos in zip(nodes, positions)]
graph.edges(edges)

graph.attr(label="D = Initial Endowment\lX = Characteristics\lY = Amount taken/given\l",
           dpi='150',
           size='3!',
           ratio='fill')

graph.render(f'img/ch_06_d_sep_02')

graph
```

```python
graph = graphviz.Digraph(format='png', engine='neato')

nodes = ['D', 'Y', 'S', 'X']
positions = ['0,0!', '3,0!', '0,3!', '3,3!']

[graph.node(n, pos=pos) for n, pos in zip(nodes, positions)]
graph.edge('D', 'Y')
graph.edge('S', 'Y', color=COLORS[1], penwidth='2')
graph.edge('X', 'Y')
graph.edge('S', 'D', color=COLORS[1], penwidth='2')

graph.attr(label="D = Initial Endowment\lS = Employment\lX = Characteristics\lY = Amount taken/given\l",
           dpi='150',
           size='3!',
           ratio='fill')

graph.render(f'img/ch_06_project_dag2')

graph
```
