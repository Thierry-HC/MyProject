```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.api import OLS
from ISLP.models import ModelSpec as MS
```

```python
df = pd.read_stata("bechtel_liesch_scheve_pnas.dta")
```

```python
X = [
    "taken", "advantagIneq", "age", "male", "GE",
    "inc_low", "inc_middle",
    "edu_low", "edu_middle",
    "left_ideology", "right_ideology",
    "employed", "unemployed", "retired", "inEducation" 
]


df2 = df.loc[
    (df["advantagIneq"] == 1) | (df["equality"] == 1),
    X
].dropna().copy()
```

```python

model = MS(df2.columns.drop("taken"), intercept=True)
D = model.fit_transform(df2)

Y = df2["taken"]
```

```python
ols = sm.OLS(Y, D).fit()

ols.summary()
```

```python
control = df2[df2["advantagIneq"] == 0]

X = sm.add_constant(control[["employed"]])
Y = control["taken"]

ols_S = sm.OLS(Y, X).fit()

ols_S.summary()
```
