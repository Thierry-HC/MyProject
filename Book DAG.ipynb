```python
import graphviz

graph = graphviz.Digraph(format='png', engine='neato')
graph.attr(overlap='false', splines='true')


# Nodes and positions
graph.node('Age', label='Age', pos='0,2!')
graph.node('Race', label='Race', pos='1.2,1.3!')
graph.node('Marital', label='Marital Status', pos='2.4,2.0!')
graph.node('Income', label='Income', pos='2.2,1.2!')
graph.node('Reading', label='Reading Habit\nGenre Preference', pos='2.5,0.2!')
graph.node('PriceAd', label='Price Ad', pos='6.2,3.2!')
graph.node('Search', label='Search', pos='5.2,1.8!')
graph.node('Purchase', label='Purchase', pos='7.0,0.8!')

# unobserved nodes: dashed border
graph.node('Privacy', label='Privacy Concern', pos='1.3,2.9!', style='dashed')
graph.node('Value', label='Perceived\nBook Value', pos='4.0,2.5!', style='dashed')

# Black edges
black_edges = [
    ('Privacy', 'Purchase'),
    ('Search', 'Purchase')
]


# Red edges
# (confounding/background paths)
red_edges = [
    ('Age', 'Purchase'),
    ('Income', 'Search'),
    ('Income', 'Purchase'),
    ('Reading', 'Search'),
    ('Reading', 'Purchase'),
    ('Marital', 'Purchase'),
    ('Age', 'Privacy'),
    ('Age', 'Marital'),
    ('Age', 'Income'),
    ('Age', 'Reading'),
    ('Race', 'Income'),
    ('Race', 'Reading'),
    ('Marital', 'Value'),
    ('Income', 'Value'),
    ('Reading', 'Value'),
    ('PriceAd', 'Value'),
    ('PriceAd', 'Search'),
    ('PriceAd', 'Purchase'),
    ('Value', 'Purchase'),
    ('Value', 'Search')
]

# draw edges
for a, b in black_edges:
    graph.edge(a, b)

for a, b in red_edges:
    graph.edge(a, b, color='red')

# save figure
graph.render('dag_price_ad', cleanup=True)
graph
```
